```python
def make_df(file_path):
    df = pd.DataFrame()

    tritrig_file = uproot.open(file_path)

    # Open ROOT tree
    tree = tritrig_file['preselection']

    # Access branches
    B_vertex = tree['vertex.']
    B_ele = tree['ele.']
    B_pos = tree['pos.']

    arr = B_vertex['vertex.pos_'].array()
    df['vertex_pos_x'] = np.array(arr['fX'])
    df['vertex_pos_y'] = np.array(arr['fY'])
    df['vertex_pos_z'] = np.array(arr['fZ'])

    df['InvM'] = np.array(B_vertex['vertex.invM_'].array())
    df['psum'] = np.array(tree['psum'].array())

    df['ele_track_n_hits'] = np.array(B_ele['ele.track_.n_hits_'].array())
    df['ele_track_d0'] = np.array(B_ele['ele.track_.d0_'].array())
    df['ele_track_phi0'] = np.array(B_ele['ele.track_.phi0_'].array())
    df['ele_track_z0'] = np.array(B_ele['ele.track_.z0_'].array())
    df['ele_track_tan_lambda'] = np.array(B_ele['ele.track_.tan_lambda_'].array())
    df['ele_track_px'] = np.array(B_ele['ele.track_.px_'].array())
    df['ele_track_py'] = np.array(B_ele['ele.track_.py_'].array())
    df['ele_track_pz'] = np.array(B_ele['ele.track_.pz_'].array())
    df['ele_track_chi2'] = np.array(B_ele['ele.track_.chi2_'].array())
    df['ele_track_x_at_ecal'] = np.array(B_ele['ele.track_.x_at_ecal_'].array())
    df['ele_track_y_at_ecal'] = np.array(B_ele['ele.track_.y_at_ecal_'].array())
    df['ele_track_z_at_ecal'] = np.array(B_ele['ele.track_.z_at_ecal_'].array())

    df['pos_track_n_hits'] = np.array(B_pos['pos.track_.n_hits_'].array())
    df['pos_track_d0'] = np.array(B_pos['pos.track_.d0_'].array())
    df['pos_track_phi0'] = np.array(B_pos['pos.track_.phi0_'].array())
    df['pos_track_z0'] = np.array(B_pos['pos.track_.z0_'].array())
    df['pos_track_tan_lambda'] = np.array(B_pos['pos.track_.tan_lambda_'].array())
    df['pos_track_px'] = np.array(B_pos['pos.track_.px_'].array())
    df['pos_track_py'] = np.array(B_pos['pos.track_.py_'].array())
    df['pos_track_pz'] = np.array(B_pos['pos.track_.pz_'].array())
    df['pos_track_chi2'] = np.array(B_pos['pos.track_.chi2_'].array())
    df['pos_track_x_at_ecal'] = np.array(B_pos['pos.track_.x_at_ecal_'].array())
    df['pos_track_y_at_ecal'] = np.array(B_pos['pos.track_.y_at_ecal_'].array())
    df['pos_track_z_at_ecal'] = np.array(B_pos['pos.track_.z_at_ecal_'].array())

    df['vertex_chi2'] = np.array(B_vertex['vertex.chi2_'].array())
    df['vtx_proj_sig'] = np.array(tree['vtx_proj_sig'].array())
    df['vtx_proj_x_sig'] = np.array(tree['vtx_proj_x_sig'].array())
    df['vtx_proj_y_sig'] = np.array(tree['vtx_proj_y_sig'].array())
    df['ele_L1_iso_significance'] = np.array(tree['ele_L1_iso_significance'].array())
    df['pos_L1_iso_significance'] = np.array(tree['pos_L1_iso_significance'].array())

    return df
```

```python
df_simp150pres = make_df('simp'+str(mass)+'pres.root')
df_bigpreselectblind = make_df('bigpreselectblind.root')
```

```python
import numpy as np

# count rows containing NaN
nan_mask = df_simp150pres.isna().any(axis=1)
n_bad = nan_mask.sum()

print("Rows containing NaN:", n_bad)

# drop them before writing pickle
df_simp150pres = df_simp150pres[~nan_mask]

# count rows containing NaN
nan_mask = df_bigpreselectblind.isna().any(axis=1)
n_bad = nan_mask.sum()

print("Rows containing NaN:", n_bad)

# drop them before writing pickle
df_bigpreselectblind = df_bigpreselectblind[~nan_mask]
```

```python
df_simp150pres.to_pickle('simp'+str(mass)+'pres.pk')
df_bigpreselectblind.to_pickle('bigpreselectblind.pk')
```
